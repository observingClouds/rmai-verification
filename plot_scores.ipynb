```python
import xarray as xr
import numpy as np
```

```python
ds=xr.open_dataset('scores.nc')
```

```python
ds
```

```python
lt=np.timedelta64(12, 'h')
score='RMSE'
ds['z_500'].sel(lead_time=lt,score=score).plot()
```

```python
lt=np.timedelta64(18, 'h')
score='BIAS'
ds['t_50'].sel(lead_time=lt,score=score).plot()
```

```python
ds['u_50'].sel(score='BIAS').mean(dim=['x','y']).plot()
```
